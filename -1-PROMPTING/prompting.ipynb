# Zero Shot Prompting
- Zero-shot prompting is a technique for large language models (LLMs) that involves providing a task or question in a prompt without any specific examples of the desired output. The model then relies on its extensive pre-trained knowledge and general understanding to infer the task and generate an appropriate response. This approach leverages the model's ability to generalize to new, unseen problems, effectively showcasing its broad knowledge base without requiring task-specific fine-tuning or demonstration data. 


```python
SYSTEM_PROMPT = "You are a math expert. Answer only math-related questions. If the query is not related to mathematics, respond with 'Sorry.'"
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {
            "role": "user",
            "content": "Explain to me how AI works"
        }
    ]
)

print(response.choices[0].message.content)
```

# Few Shot Prompting
- Few-shot prompting is a technique used to improve large language models (LLMs) performance by including a small number of examples within the prompt itself, demonstrating the desired task and format. This approach guides the model by providing context and examples for a specific task, helping it to generate more accurate and relevant outputs without requiring large amounts of labeled data or updating the model's weights.  


```python
SYSTEM_PROMPT_1 = """
You are a math expert. Answer only math-related questions. If the query is not related to mathematics, respond with 'Sorry.'

Examples:
Q: What is the derivative of x²?
A: The derivative of x² is 2x.

Q: Solve for x: 2x + 3 = 7
A: Subtract 3 from both sides: 2x = 4. Then divide by 2: x = 2.

Q: Who won the World Cup in 2022?
A: Sorry I am trained to answer only on Math-Related Questions

Q: What is the integral of sin(x)?
A: The integral of sin(x) is -cos(x) + C.

Q: Tell me a joke.
A: Sorry I am trained to answer only on Math-Related Questions

Now respond to the next query accordingly.
"""
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT_1},
        {
            "role": "user",
            "content": "Explain to me how AI works"
        }
    ]
)

print(response.choices[0].message.content)
```

# Chain of Thought 
- 🔗 Chain of Thought (CoT) prompting is a technique in AI that guides models to reason step-by-step before giving an answer. Instead of jumping straight to a conclusion, the model is encouraged to "think aloud" — breaking down the problem into intermediate steps that lead to a final solution.
🧠 Why It Matters
- It improves accuracy, especially for complex tasks like math, logic, and commonsense reasoning.
- It makes the model’s reasoning transparent, so you can follow how it arrived at the answer.
- It mimics how humans solve problems — by working through them logically.



```python
SYSTEM_PROMPT_2 = """
You are a math expert. Answer only math-related questions. If the query is not related to mathematics, respond with 'Sorry.' For math questions, explain your reasoning step-by-step before giving the final answer.

Examples:
Q: What is the derivative of x²?
A: Let's think step-by-step.
The derivative of xⁿ with respect to x is n·xⁿ⁻¹.
Here, n = 2, so the derivative of x² is 2x.
Final answer: 2x.

Q: Solve for x: 2x + 3 = 7
A: Let's think step-by-step.
First, subtract 3 from both sides: 2x = 4.
Then divide both sides by 2: x = 2.
Final answer: x = 2.

Q: Who won the World Cup in 2022?
A: Sorry, I am trained to answer only math-related questions.

Q: What is the integral of sin(x)?
A: Let's think step-by-step.
The integral of sin(x) with respect to x is -cos(x), plus a constant of integration.
Final answer: -cos(x) + C.

Q: Tell me a joke.
A: Sorry, I am trained to answer only math-related questions.

Now respond to the next query accordingly.
"""
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT_2},
        {
            "role": "user",
            "content": "Explain right angled traingle"
        }
    ]
)

print(response.choices[0].message.content)
```

# Persona Based Prompting
- Persona-Based Prompting is a technique where you assign a specific identity or role to an AI model to guide its behavior, tone, and expertise. This helps the model generate responses that are more relevant, consistent, and tailored to the task at hand.


```python
persona_prompt = """
You are Dr. Maya Chen, a leading UX researcher with 15 years of experience in fintech applications.
Your expertise includes analyzing user behavior patterns and identifying usability issues in financial interfaces.
You communicate clearly using data-backed insights and avoid technical jargon.
Focus only on mobile banking application design challenges and do not provide general UX advice outside this domain.
"""
```

```python
response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {"role": "system", "content": persona_prompt},
        {
            "role": "user",
            "content": "Explain right angled traingle"
        }
    ]
)

print(response.choices[0].message.content)
```
