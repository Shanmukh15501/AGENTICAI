# 🧠 Large Language Models (LLMs)

## 🔹 What are LLMs?
- **LLM = Large Language Model**
- Trained to understand and generate **human language**
- Examples: **ChatGPT, Gemini, Claude**
- Built on **Transformer architecture** (introduced in Google’s paper *"Attention Is All You Need"* 2017)

---

## 🔹 Key Idea of Transformers
- Originally used for **machine translation** (predicting next word in another language)
- Today extended to chatbots, summarization, coding assistants, etc.

---

## 🔹 How LLMs Work
1. **User input** → `"Hello, how are you?"`
2. **Tokenization**  
   - Input is split into smaller chunks (**tokens**)  
   - Each token is mapped to a **unique number** (depends on the model’s vocabulary)  

Example:
    "Hello" → 15496
    "," → 11
    "how" → 703
    "are" → 389
    "you" → 345
    "?" → 30

So input = `[15496, 11, 703, 389, 345, 30]`

3. **Model prediction**
- Tokens are processed by the LLM
- The model predicts the **most likely next token(s)**

4. **Output tokens**
- Example: `[366, 345, 318, 12]`

5. **Detokenization**
- Convert back to text: `"I am fine."`

---

## 🔹 Why Tokens?
- Computers don’t understand words directly.
- Tokens are a **numerical representation** of language.
- Different models (GPT, Gemini, Claude) have different vocabularies & tokenizers.

```python
import tiktoken

# Load the tokenizer for a GPT model (e.g., GPT-3.5 / GPT-4)
enc = tiktoken.encoding_for_model("gpt-3.5-turbo")

# Input text
text = "Hello, how are you?"

# Encode (text → tokens)
tokens = enc.encode(text)
print("Token IDs:", tokens)

# Decode (tokens → text)
decoded_text = enc.decode(tokens)
print("Decoded text:", decoded_text)

# Show tokens with their string values
print("\nDetailed breakdown:")
for token in tokens:
    print(f"{token} → {enc.decode([token])}")
```

### Vector Embeddings
#### What is an embedding?
- An embedding is a way to represent text (words, sentences, documents) as a list of numbers (vector).
 
- Instead of raw words, models use embeddings because:

- Numbers are easier for machines to process.

- Similar meanings → closer vectors in space.

- Think of it like mapping words into a multi-dimensional space where similar words are near each other.
- Vector embeddings are numerical representations of data points—like words, images, or audio—transformed into arrays of numbers (vectors) that computers can process and understand. These vectors capture the meaning and relationships within the data, representing similar data points with vectors positioned close to each other in a multidimensional space. This process allows machine learning models to find patterns, perform tasks like sentiment analysis or image similarity, and power applications such as recommendation systems and advanced search.  

### Positional Encoding in Transformers
#### Problem

- Transformers don’t process words sequentially (like RNNs or LSTMs).

-  They read the whole sentence in parallel.

- But then… how do they know the order of words?

* "I am eating an apple"
* "Apple eating am I"
* Same words, but different meaning → order matters.


#### Solution: Positional Encoding

- Add position information to each token’s embedding.
- This tells the model where a word appears in the sequence.
- The encoding is added to the word embedding before feeding into the Transformer layers.
- Input vector = WordEmbedding + PositionalEncoding
